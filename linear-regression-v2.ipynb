```python
import numpy as np
from numpy.typing import NDArray
import matplotlib.pyplot as plt

from linear_regression import LinearRegression
from utils import *
```

## Training

```python
ex_data_file = "data/ex1data1.txt"
X_train, y_train = load_data(ex_data_file)
print ('The shape of x_train is:', X_train.shape)
print ('The shape of y_train is: ', y_train.shape)
print ('Number of training examples (m):', len(X_train))

w_init = np.zeros(1)
b_init = 0

alpha = 1.0e-2
iters_num = 1500

print('Start training...')
model = LinearRegression(X_train, y_train)
w, b, j_history, p_history = model.fit(w_init, b_init, alpha, iters_num)

print(f'w = {w[0]:8.4f}, b = {b:8.4f}')

fig, ax1 = plt.subplots(1, 1, constrained_layout=True, figsize=(12,4))

ax1.set_title("Cost vs. iteration(start)");  
ax1.set_ylabel('Cost') 
ax1.set_xlabel('iteration step')
ax1.plot(j_history)

plt.show()
```

```python
fig, ax = plt.subplots(1, 1, constrained_layout=True, figsize=(12,4))
ax.scatter(X_train, y_train, c="r")
y_pred = model.predict(X_train)
ax.plot(X_train, y_pred, c="b")
plt.show()
```
